```python
import uproot
import matplotlib.pyplot as plt
import mplhep as hep

file = uproot.open('/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_GUNDAM/data/Asimov/asimov_dpT.root')

# Check the category histograms
categories = [0, 1, 3, 4]

print("="*70)
print("CHECKING CATEGORY HISTOGRAMS:")
print("="*70)

for cat in categories:
    path = f"FitterEngine/postFit/samples/histograms/signal_deltapT/reco_dpT/category/{cat}/MC_TH1D"
    try:
        hist = file[path]
        values, edges = hist.to_numpy()
        entries = hist.member('fEntries')

        print(f"\nCategory {cat}:")
        print(f"  Entries: {entries}")
        print(f"  Bins: {len(values)}")
        print(f"  Edge range: [{edges[0]:.3f}, {edges[-1]:.3f}]")
        print(f"  Values sum: {values.sum():.3f}")

        if len(values) > 1 and edges[-1] > 0.1:
            print(f"  ✓ GOOD! This has proper binning")

    except Exception as e:
        print(f"Category {cat}: ERROR - {e}")

# Now try plotting one that works
print("\n" + "="*70)
print("ATTEMPTING TO PLOT:")
print("="*70)

fig, ax = plt.subplots(figsize=(10, 6))

# Stack all categories
for cat in categories:
    path = f"FitterEngine/postFit/samples/histograms/signal_deltapT/reco_dpT/category/{cat}/Data_TH1D"
    try:
        hist = file[path]
        values, edges = hist.to_numpy()
        if len(values) > 1 and edges[-1] > 0.1:
            hep.histplot(hist, ax=ax, label=f"Category {cat}", stack=True)
    except:
        pass

ax.set_xlabel("$\Delta p_T$ [GeV/c]")
ax.set_ylabel("Events")
ax.legend()
plt.tight_layout()
plt.show()

print("\nPlot saved as: gundam_plot.png")
```
