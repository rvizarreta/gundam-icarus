```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import mplhep as hep

file = uproot.open('/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_GUNDAM/data/Asimov/prefit_asimov_dpT.root')

# Category configuration - merge all signal into one
categories_config = {
    0: ('Signal', '#1a7000'),      # kGreen+2 (dark)
    3: ('Signal', '#1a7000'),      # kGreen+2 (dark)
    1: ('Signal', '#1a7000'),      # Use SAME color for all signal
    4: ('Signal', '#1a7000'),      # Use SAME color for all signal
    6: ('Other CC', '#ff9933'),    # kOrange+7
    7: ('Other CC', '#ff9933'),    # kOrange+7
    8: ('Other CC', '#ff9933'),    # kOrange+7
    9: (r'$\nu$ NC', '#cc0000'),   # kRed
    2: ('OOFV', '#cc6600'),        # kOrange-3
    5: ('OOFV', '#cc6600'),        # kOrange-3
}

# Stacking order (backgrounds first, signal last)
stack_order = [2, 5, 9, 6, 7, 8, 1, 4, 0, 3]

# Extract all category histograms
category_data = {}
bin_edges = None

for cat in categories_config.keys():
    path = f"toyGen/plots/histograms/signal_dpT (pre-fit)/reco_dpT/category/{cat}/MC_TH1D"
    try:
        hist = file[path]
        values, edges = hist.to_numpy()

        if bin_edges is None:
            bin_edges = edges

        category_data[cat] = values
        label, color = categories_config[cat]

    except KeyError:
        print(f"Category {cat}: NOT FOUND")

# Extract data
data_hist = file["toyGen/plots/histograms/signal_dpT (pre-fit)/reco_dpT/Data_TH1D"]
data_values, data_edges = data_hist.to_numpy()

# Group categories by (label, color) - this will merge all signal
unique_categories = {}
for cat in stack_order:
    if cat not in category_data:
        continue

    label, color = categories_config[cat]
    key = (label, color)

    if key not in unique_categories:
        unique_categories[key] = np.zeros_like(category_data[cat])

    unique_categories[key] += category_data[cat]

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8),
                                gridspec_kw={'height_ratios': [3, 1], 'hspace': 0.05})

# Top panel: Stacked histogram
cumulative = np.zeros_like(list(unique_categories.values())[0])

for (label, color), values in unique_categories.items():
    # Plot filled area
    ax1.fill_between(
        bin_edges[:-1],
        cumulative,
        cumulative + values,
        step='post',
        label=label,
        color=color,
        alpha=0.8,
        linewidth=0
    )
    # Last bin
    ax1.fill_between(
        [bin_edges[-2], bin_edges[-1]],
        [cumulative[-1], cumulative[-1]],
        [cumulative[-1] + values[-1], cumulative[-1] + values[-1]],
        color=color,
        alpha=0.8,
        linewidth=0
    )
    cumulative += values

# Data points
data_centers = (data_edges[:-1] + data_edges[1:]) / 2
data_errors = np.sqrt(data_values)
ax1.errorbar(
    data_centers,
    data_values,
    yerr=data_errors,
    fmt='o',
    color='black',
    markersize=5,
    label='Data',
    linewidth=1.2,
    capsize=2
)

ax1.set_ylabel('Events', fontsize=12)
ax1.set_xlim(bin_edges[0], bin_edges[-1])
ax1.set_ylim(0, ax1.get_ylim()[1] * 1.3)
ax1.tick_params(labelbottom=False)
handles, labels = ax1.get_legend_handles_labels()
ax1.legend(handles[::-1], labels[::-1], loc='upper right', fontsize=10)
ax1.set_title('ICARUS CC0π: Pre-Fit', fontsize=14, pad=10)
ax1.grid(True, alpha=0.3, linestyle='--', axis='y')

# Bottom panel: Data/MC ratio
ratio = data_values / cumulative
ratio_err = data_errors / cumulative

ax2.errorbar(
    data_centers,
    ratio,
    yerr=ratio_err,
    fmt='o',
    color='black',
    markersize=5,
    linewidth=1.2,
    capsize=2
)
ax2.axhline(y=1, color='gray', linestyle='--', linewidth=1)
ax2.fill_between(bin_edges, 0.9, 1.1, alpha=0.2, color='gray', step='post')
ax2.set_xlabel(r'$\delta p_T$ [MeV/c]', fontsize=12)
ax2.set_ylabel('Data/MC', fontsize=11)
ax2.set_xlim(bin_edges[0], bin_edges[-1])
ax2.set_ylim(0.5, 1.5)
ax2.grid(True, alpha=0.3, linestyle='--')

plt.show()
```
