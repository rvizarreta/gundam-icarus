```python
import uproot
import matplotlib.pyplot as plt
import mplhep as hep
import numpy as np

file = uproot.open('/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_GUNDAM/data/Asimov/asimov_dpT.root')

# ROOT color codes to matplotlib hex colors
root_to_mpl = {
    418: '#1a7000',  # kGreen+2 (dark green)
    416: '#00a000',  # kGreen (lighter green)
    807: '#ff9933',  # kOrange+7 (orange)
    632: '#cc0000',  # kRed (red)
    797: '#cc6600',  # kOrange-3 (brown)
    880: '#9966cc',  # kViolet (purple)
    1: '#000000',    # kBlack (black)
}

# Category configuration matching your YAML
# Format: {interaction_type: (label, color_code)}
categories = {
    0: ('Signal', 418),      # Signal (dark green)
    3: ('Signal', 418),      # Signal (dark green)
    1: ('Signal', 416),      # Signal (light green)
    4: ('Signal', 416),      # Signal (light green)
    6: ('Other CC', 807),    # Other CC (orange)
    7: ('Other CC', 807),    # Other CC (orange)
    8: ('Other CC', 807),    # Other CC (orange)
    9: (r'$\nu$ NC', 632),   # ν NC (red)
    2: ('OOFV', 797),        # OOFV (brown)
    5: ('OOFV', 797),        # OOFV (brown)
    10: ('Cosmic', 880),     # Cosmic (purple)
}

# Stacking order (backgrounds first, signal last)
stack_order = [10, 2, 5, 9, 6, 7, 8, 1, 4, 0, 3]

# Storage for histograms
hist_data = {}
bin_edges = None

# Read all category histograms
print("Reading histograms...")
for cat_type in categories.keys():
    path = f"FitterEngine/preFit/plots/histograms/signal_dpT/reco_dpT/category/{cat_type}/MC_TH1D"
    try:
        hist = file[path]
        values, edges = hist.to_numpy()

        if bin_edges is None:
            bin_edges = edges

        label, color_code = categories[cat_type]
        hist_data[cat_type] = {
            'values': values,
            'label': label,
            'color': root_to_mpl[color_code]
        }
        print(f"  Category {cat_type} ({label}): {values.sum():.1f} events")

    except KeyError:
        print(f"  Category {cat_type}: NOT FOUND")

# Create the plot
fig, ax = plt.subplots(figsize=(10, 7))

# Prepare data for stacking
# Group by unique (label, color) combinations
unique_categories = {}
for cat_type in stack_order:
    if cat_type not in hist_data:
        continue

    label = hist_data[cat_type]['label']
    color = hist_data[cat_type]['color']
    key = (label, color)

    if key not in unique_categories:
        unique_categories[key] = np.zeros_like(hist_data[cat_type]['values'])

    unique_categories[key] += hist_data[cat_type]['values']

# Plot stacked histogram
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
cumulative = np.zeros_like(list(unique_categories.values())[0])

for (label, color), values in unique_categories.items():
    ax.fill_between(
        bin_edges[:-1],
        cumulative,
        cumulative + values,
        step='post',
        label=label,
        color=color,
        alpha=0.8,
        linewidth=0
    )
    # Add edge to last bin
    ax.fill_between(
        [bin_edges[-2], bin_edges[-1]],
        [cumulative[-1], cumulative[-1]],
        [cumulative[-1] + values[-1], cumulative[-1] + values[-1]],
        color=color,
        alpha=0.8,
        linewidth=0
    )
    cumulative += values

# Read and overlay data
try:
    data_hist = file["FitterEngine/preFit/plots/histograms/signal_dpT/reco_dpT/Data_TH1D"]
    data_values, data_edges = data_hist.to_numpy()
    data_errors = np.sqrt(data_values)

    # Plot data as points with error bars
    data_centers = (data_edges[:-1] + data_edges[1:]) / 2
    ax.errorbar(
        data_centers,
        data_values,
        yerr=data_errors,
        fmt='o',
        color='black',
        markersize=6,
        label='Data',
        linewidth=1.5,
        capsize=3
    )
    print(f"\nData: {data_values.sum():.1f} events")

except KeyError:
    print("\nData histogram not found")

# Styling
ax.set_xlabel(r'$\Delta p_T$ [GeV/c]', fontsize=14)
ax.set_ylabel('Events', fontsize=14)
ax.set_xlim(bin_edges[0], bin_edges[-1])
ax.set_ylim(0, ax.get_ylim()[1] * 1.2)  # Add 20% headroom

# Legend (reverse order so signal at top)
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[::-1], labels[::-1], loc='upper right', fontsize=11, framealpha=0.9)

ax.set_title('ICARUS CC0π: Pre-Fit', fontsize=15, pad=10)
ax.grid(True, alpha=0.3, linestyle='--')

plt.show()

# Print summary
print("\n" + "="*70)
print("SUMMARY:")
print("="*70)
total_mc = sum(values.sum() for values in unique_categories.values())
print(f"Total MC: {total_mc:.1f} events")
try:
    print(f"Total Data: {data_values.sum():.1f} events")
    print(f"Data/MC ratio: {data_values.sum()/total_mc:.3f}")
except:
    pass
```
