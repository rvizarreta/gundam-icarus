```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import glob
import pandas as pd


def _read_nuisflat_enu(file_dir, bin_edges, flux_binwidth_divided=True):
    """
    Read all nuisflat files from one directory for Enu comparison.
    Returns (h_weighted, flux_integral, n_files).
    """
    files = sorted(glob.glob(f"{file_dir}/output_GENIE_*.nuisflat.root"))
    if not files:
        raise FileNotFoundError(f'No files found in {file_dir}')

    df_all = None
    flux_integral = None

    for f_path in files:
        with uproot.open(f_path) as f:
            df = f["FlatTree_VARS"].arrays(
                ["Enu_true", "InputWeight", "fScaleFactor", "ICARUS_1muNp0pi_IsSignal"],
                library="pd"
            )
            if df_all is None:
                df_all = df
                flux_hist = f["FlatTree_FLUX"].to_numpy()
                flux_integral = (
                    np.sum(flux_hist[0] * np.diff(flux_hist[1]))
                    if flux_binwidth_divided
                    else np.sum(flux_hist[0])
                )
                flux_integral *= 1e-4  # /m² → /cm²
            else:
                df_all = pd.concat([df_all, df], ignore_index=True)

    df_all["FinalWeight"] = df_all["InputWeight"] * df_all["fScaleFactor"]
    df_all["Enu_true_GeV"] = df_all["Enu_true"] * 1e-3
    df_sig = df_all.query("ICARUS_1muNp0pi_IsSignal == 1")

    h = np.histogram(df_sig["Enu_true_GeV"], bins=bin_edges,
                     weights=df_sig["FinalWeight"])[0]

    print(f"[NUISANCE] N total events:  {len(df_all)}")
    print(f"[NUISANCE] N signal events: {len(df_sig)}")

    return h, flux_integral, len(files)


def compare_enu_caf_vs_nuisance(
    caf_file_path,
    nuisance_file_dir,
    nuisance_file_dir_numubar=None,
    caf_tree_path="events/nominal/signal",
    bin_edges=np.linspace(0, 5, 51),
    normalize=True,
    label_caf_numu=r"Signal Tree ($\nu_\mu$)",
    label_caf_numubar=r"Signal Tree ($\bar{\nu}_\mu$)",
    label_numu=r"GENIE NUISANCE ($\nu_\mu$)",
    label_numubar=r"GENIE NUISANCE ($\bar{\nu}_\mu$)",
    figsize=(7, 6),
):
    bin_edges  = np.array(bin_edges)
    bin_widths = np.diff(bin_edges)

    # ── 1. Read CAF Enu + PDG ────────────────────────────────────────────────
    with uproot.open(caf_file_path) as f:
        tree = f[caf_tree_path]
        enu_caf = tree["true_neutrino_energy"].array(library="np")
        pdg_caf = tree["true_pdg"].array(library="np")
        cat_caf = tree["category"].array(library="np")

    cat_mask = np.isin(cat_caf, [0, 1, 3, 4])
    enu_caf = enu_caf[cat_mask]
    pdg_caf = pdg_caf[cat_mask]

    mask_numu    = pdg_caf ==  14
    mask_numubar = pdg_caf == -14

    print(f"[CAF] N events total:   {len(enu_caf)}")
    print(f"[CAF] N numu    (14):   {mask_numu.sum()}")
    print(f"[CAF] N numubar (-14):  {mask_numubar.sum()}")

    # ── 2. Read NUISANCE numu ────────────────────────────────────────────────
    h_numu, flux_numu, n_files_numu = _read_nuisflat_enu(nuisance_file_dir, bin_edges)
    print(f"[NUISANCE numu]    {n_files_numu} files, flux integral = {flux_numu:.4e} /cm²")
    print(f"[NUISANCE numu]    N signal events (weighted sum): {h_numu.sum():.2f}")

    # ── 3. Read NUISANCE numubar (optional) ──────────────────────────────────
    if nuisance_file_dir_numubar is not None:
        h_numubar, flux_numubar, n_files_numubar = _read_nuisflat_enu(
            nuisance_file_dir_numubar, bin_edges
        )
        print(f"[NUISANCE numubar] {n_files_numubar} files, flux integral = {flux_numubar:.4e} /cm²")

    # ── 4. Histogram CAF split by PDG ────────────────────────────────────────
    h_caf_numu,          _ = np.histogram(enu_caf[mask_numu], bins=bin_edges)
    h_caf_numu_weighted, _ = np.histogram(enu_caf[mask_numu], bins=bin_edges)
    h_caf_numubar,       _ = np.histogram(enu_caf[mask_numubar], bins=bin_edges)

    h_numu_plot = h_numu / n_files_numu
    if nuisance_file_dir_numubar is not None:
        h_numubar_plot = h_numubar / n_files_numubar

    # ── 5. Normalize ─────────────────────────────────────────────────────────
    if normalize:
        h_caf_numu          = h_caf_numu          / (h_caf_numu.sum()          * bin_widths)
        h_caf_numu_weighted = h_caf_numu_weighted / (h_caf_numu_weighted.sum() * bin_widths)
        h_caf_numubar       = h_caf_numubar       / (h_caf_numubar.sum()       * bin_widths)
        h_numu_plot         = h_numu_plot          / (h_numu_plot.sum()         * bin_widths)
        if nuisance_file_dir_numubar is not None:
            h_numubar_plot = h_numubar_plot / (h_numubar_plot.sum() * bin_widths)

    # ── 6. Plot ──────────────────────────────────────────────────────────────
    fig_numu, ax_numu = plt.subplots(figsize=figsize)

    ax_numu.step(bin_edges, np.append(h_caf_numu, h_caf_numu[-1]),
                 where="post", label=label_caf_numu, linewidth=1.8, color='C0')
    ax_numu.step(bin_edges, np.append(h_numu_plot, h_numu_plot[-1]),
                 where="post", label=label_numu, linewidth=1.8, color='C1')

    ax_numu.set_xlabel(r"$E_\nu^{\rm true}$ [GeV]", fontsize=13)
    ax_numu.set_xlim(bin_edges[0], bin_edges[-1])
    ax_numu.set_ylim(bottom=0)
    ax_numu.legend(fontsize=11)

    if nuisance_file_dir_numubar is not None:
        fig_numubar, ax_numubar = plt.subplots(figsize=figsize)

        ax_numubar.step(bin_edges, np.append(h_caf_numubar, h_caf_numubar[-1]),
                        where="post", label=label_caf_numubar, linewidth=1.8, color='C0')
        ax_numubar.step(bin_edges, np.append(h_numubar_plot, h_numubar_plot[-1]),
                        where="post", label=label_numubar, linewidth=1.8, color='C1')

        ax_numubar.set_xlabel(r"$E_\nu^{\rm true}$ [GeV]", fontsize=13)
        ax_numubar.set_xlim(bin_edges[0], bin_edges[-1])
        ax_numubar.set_ylim(bottom=0)
        ax_numubar.legend(fontsize=11)

        return fig_numu, ax_numu, fig_numubar, ax_numubar

    return fig_numu, ax_numu
```

```python
import numpy as np
def compare_dpt_caf_vs_nuisance(
    caf_file_path,
    nuisance_file_dir,
    nuisance_file_dir_numubar=None,
    caf_tree_path="events/full/signal",
    bin_edges=np.linspace(0, 1, 51),
    normalize=True,
    label_caf_numu=r"Signal Tree ($\nu_\mu$)",
    label_caf_numubar=r"Signal Tree ($\bar{\nu}_\mu$)",
    label_numu=r"GENIE NUISANCE ($\nu_\mu$)",
    label_numubar=r"GENIE NUISANCE ($\bar{\nu}_\mu$)",
    figsize=(7, 6),
):
    bin_edges  = np.array(bin_edges)
    bin_widths = np.diff(bin_edges)

    # ── 1. Read CAF true_dpT + PDG ───────────────────────────────────────────
    with uproot.open(caf_file_path) as f:
        tree = f[caf_tree_path]
        dpt_caf = tree["true_dpT"].array(library="np")
        pdg_caf = tree["true_pdg"].array(library="np")
        cat_caf = tree["category"].array(library="np")

    cat_mask = np.isin(cat_caf, [0, 1, 3, 4])
    dpt_caf = tree["true_dpT"].array(library="np") * 1e-3  # MeV/c → GeV/c
    pdg_caf  = pdg_caf[cat_mask]

    mask_numu    = pdg_caf ==  14
    mask_numubar = pdg_caf == -14

    print(f"[CAF] N events total:   {len(dpt_caf)}")
    print(f"[CAF] N numu    (14):   {mask_numu.sum()}")
    print(f"[CAF] N numubar (-14):  {mask_numubar.sum()}")

    # ── 2. Read NUISANCE numu δpT ────────────────────────────────────────────
    h_numu, n_files_numu = _read_nuisflat_dpt(nuisance_file_dir, bin_edges)
    print(f"[NUISANCE numu]    {n_files_numu} files")
    print(f"[NUISANCE numu]    N signal events (weighted sum): {h_numu.sum():.2f}")

    # ── 3. Read NUISANCE numubar δpT (optional) ──────────────────────────────
    if nuisance_file_dir_numubar is not None:
        h_numubar, n_files_numubar = _read_nuisflat_dpt(nuisance_file_dir_numubar, bin_edges)
        print(f"[NUISANCE numubar] {n_files_numubar} files")

    # ── 4. Histogram CAF split by PDG ────────────────────────────────────────
    h_caf_numu,    _ = np.histogram(dpt_caf[mask_numu],    bins=bin_edges)
    h_caf_numubar, _ = np.histogram(dpt_caf[mask_numubar], bins=bin_edges)

    h_numu_plot = h_numu / n_files_numu
    if nuisance_file_dir_numubar is not None:
        h_numubar_plot = h_numubar / n_files_numubar

    # ── 5. Normalize ─────────────────────────────────────────────────────────
    if normalize:
        h_caf_numu    = h_caf_numu    / (h_caf_numu.sum()    * bin_widths)
        h_caf_numubar = h_caf_numubar / (h_caf_numubar.sum() * bin_widths)
        h_numu_plot   = h_numu_plot   / (h_numu_plot.sum()   * bin_widths)
        if nuisance_file_dir_numubar is not None:
            h_numubar_plot = h_numubar_plot / (h_numubar_plot.sum() * bin_widths)

    # ── 6. Plot ──────────────────────────────────────────────────────────────
    fig_numu, ax_numu = plt.subplots(figsize=figsize)

    ax_numu.step(bin_edges, np.append(h_caf_numu, h_caf_numu[-1]),
                 where="post", label=label_caf_numu, linewidth=1.8, color='C0')
    ax_numu.step(bin_edges, np.append(h_numu_plot, h_numu_plot[-1]),
                 where="post", label=label_numu, linewidth=1.8, color='C1')

    ax_numu.set_xlabel(r"$\delta p_T$ [GeV/c]", fontsize=13)
    ax_numu.set_xlim(bin_edges[0], bin_edges[-1])
    ax_numu.set_ylim(bottom=0)
    ax_numu.legend(fontsize=11)

    if nuisance_file_dir_numubar is not None:
        fig_numubar, ax_numubar = plt.subplots(figsize=figsize)

        ax_numubar.step(bin_edges, np.append(h_caf_numubar, h_caf_numubar[-1]),
                        where="post", label=label_caf_numubar, linewidth=1.8, color='C0')
        ax_numubar.step(bin_edges, np.append(h_numubar_plot, h_numubar_plot[-1]),
                        where="post", label=label_numubar, linewidth=1.8, color='C1')

        ax_numubar.set_xlabel(r"$\delta p_T$ [GeV/c]", fontsize=13)
        ax_numubar.set_xlim(bin_edges[0], bin_edges[-1])
        ax_numubar.set_ylim(bottom=0)
        ax_numubar.legend(fontsize=11)

        return fig_numu, ax_numu, fig_numubar, ax_numubar

    return fig_numu, ax_numu


def _read_nuisflat_dpt(file_dir, bin_edges):
    """
    Read all nuisflat files from one directory for δpT comparison.
    Returns (h_weighted, n_files).
    Note: no flux normalization needed since we're shape-comparing only.
    """
    files = sorted(glob.glob(f"{file_dir}/output_GENIE_*.nuisflat.root"))
    if not files:
        raise FileNotFoundError(f'No files found in {file_dir}')

    df_all = None

    for f_path in files:
        with uproot.open(f_path) as f:
            df = f["FlatTree_VARS"].arrays(
                ["ICARUS_1muNp0pi_deltaPT", "InputWeight", "fScaleFactor",
                 "ICARUS_1muNp0pi_IsSignal"],
                library="pd"
            )
            df_all = df if df_all is None else pd.concat([df_all, df], ignore_index=True)

    df_all["FinalWeight"] = df_all["InputWeight"] * df_all["fScaleFactor"]
    df_sig = df_all.query("ICARUS_1muNp0pi_IsSignal == 1")

    print(f"[NUISANCE] N total events:  {len(df_all)}")
    print(f"[NUISANCE] N signal events: {len(df_sig)}")

    # Convert MeV → GeV if needed (check your NUISANCE output units)
    dpt_vals = df_sig["ICARUS_1muNp0pi_deltaPT"]

    h = np.histogram(dpt_vals, bins=bin_edges, weights=df_sig["FinalWeight"])[0]

    return h, len(files)
```

```python
"""
Extract POT from events/full/POT TH1D — exhaustively.

The earlier attempts failed with 'len() of unsized object' which suggests
fN is a single scalar in some uproot version, or the type interpretation
is unusual. This script tries every plausible extraction path and prints
all the raw members so we can see exactly what's in the histogram.
"""

import uproot
import numpy as np

OLD_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/icarus_numi_numu_mc_onbeam_offbeam_syst_oficial.root"
NEW_PATH = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/1muNp0pi_Nge1_uncontained_systematics.root"


def deep_inspect_pot(path, label):
    print(f"\n{'='*70}")
    print(f"{label}: {path.split('/')[-1]}")
    print('='*70)

    f = uproot.open(path)
    h = f["events/full/POT"]

    print(f"\nObject type: {type(h)}")
    print(f"classname:   {h.classname if hasattr(h, 'classname') else 'n/a'}")

    # Method 1: Print every member with full contents
    print(f"\n--- Method 1: all_members ---")
    members = h.all_members
    for k, v in members.items():
        # Try to print the value robustly
        try:
            if hasattr(v, "__len__"):
                try:
                    length = len(v)
                    if length < 30:
                        print(f"  {k:25s} = {v}   (len={length})")
                    else:
                        arr = np.asarray(v)
                        print(f"  {k:25s} : array of len {length}, "
                              f"sum={arr.sum():.4e}, "
                              f"nonzero={np.count_nonzero(arr)}, "
                              f"first 5: {arr[:5]}")
                except TypeError:
                    print(f"  {k:25s} = {v}   (has __len__ but len() failed)")
            else:
                print(f"  {k:25s} = {v}")
        except Exception as e:
            print(f"  {k:25s} : <error: {e}>")

    # Method 2: Try uproot's high-level histogram interface
    print(f"\n--- Method 2: uproot histogram interface ---")
    try:
        values = h.values()
        print(f"  h.values():        {values}  type={type(values)}, shape={getattr(values, 'shape', 'n/a')}")
        if hasattr(values, "sum"):
            print(f"  h.values().sum():  {values.sum():.4e}")
    except Exception as e:
        print(f"  h.values() failed: {e}")

    try:
        edges = h.axis().edges()
        print(f"  h.axis().edges():  {edges}  shape={getattr(edges, 'shape', 'n/a')}")
    except Exception as e:
        print(f"  h.axis().edges() failed: {e}")

    try:
        np_hist = h.to_numpy()
        print(f"  h.to_numpy():      counts={np_hist[0]}, edges={np_hist[1]}")
        print(f"  counts.sum():      {np.asarray(np_hist[0]).sum():.4e}")
    except Exception as e:
        print(f"  h.to_numpy() failed: {e}")

    try:
        boost_hist = h.to_boost()
        print(f"  h.to_boost():      {boost_hist}")
        print(f"  boost view:        {boost_hist.view()}")
        print(f"  boost sum:         {boost_hist.view().sum():.4e}")
    except Exception as e:
        print(f"  h.to_boost() failed: {e}")

    # Method 3: fEntries × fTsumw arithmetic
    print(f"\n--- Method 3: derived quantities ---")
    fEntries = members.get("fEntries")
    fTsumw   = members.get("fTsumw")
    fTsumw2  = members.get("fTsumw2")
    fTsumwx  = members.get("fTsumwx")
    print(f"  fEntries  = {fEntries}")
    print(f"  fTsumw    = {fTsumw}     (sum of weights — often == POT)")
    print(f"  fTsumw2   = {fTsumw2}    (sum of weights^2)")
    print(f"  fTsumwx   = {fTsumwx}    (sum of w*x)")
    if fEntries and fTsumw:
        print(f"  fTsumw/fEntries = {fTsumw/fEntries:.4e}  (mean weight per entry)")

    # Method 4: Look at fN array carefully
    print(f"\n--- Method 4: fN bin contents inspection ---")
    fN = members.get("fN")
    if fN is None:
        print("  fN is None")
    else:
        print(f"  fN type: {type(fN)}")
        try:
            arr = np.asarray(fN)
            print(f"  asarray shape: {arr.shape}, dtype: {arr.dtype}")
            print(f"  asarray ndim: {arr.ndim}")
            if arr.ndim == 0:
                print(f"  scalar value: {arr.item()}")
            else:
                print(f"  total sum (incl underflow/overflow): {arr.sum():.4e}")
                if len(arr) >= 3:
                    print(f"  underflow (bin 0): {arr[0]}")
                    print(f"  overflow  (bin -1): {arr[-1]}")
                    print(f"  in-range sum:      {arr[1:-1].sum():.4e}")
                if len(arr) <= 20:
                    print(f"  full contents:     {arr}")
                else:
                    print(f"  first 10:          {arr[:10]}")
                    print(f"  last 10:           {arr[-10:]}")
        except Exception as e:
            print(f"  asarray failed: {e}")
            # Try iterating
            try:
                vals = list(fN)
                print(f"  Iterated: len={len(vals)}, first 5: {vals[:5]}")
            except Exception as e2:
                print(f"  Iteration also failed: {e2}")

    # Method 5: Just dump the raw bytes via streamerinfo if all else fails
    print(f"\n--- Method 5: TTree-style read (in case POT is actually a small tree) ---")
    try:
        if hasattr(h, "arrays"):
            arr = h.arrays(library="np")
            print(f"  Got arrays: {arr}")
    except Exception as e:
        print(f"  Not a TTree: {e}")


for path, label in [(OLD_PATH, "OLD MC"), (NEW_PATH, "NEW MC")]:
    deep_inspect_pot(path, label)

# Summary at the end
print(f"\n\n{'='*70}\nSUMMARY\n{'='*70}")
print("Documented POT:")
print(f"  OLD: 1.47115e+20")
print(f"  NEW: 7.78521e+20")
print(f"  ratio NEW/OLD: {7.78521/1.47115:.4f}")
print(f"\nIf the deficit is purely a POT bookkeeping error,")
print(f"  actual NEW POT should be: 7.78521e20 × 0.9167 = {7.78521e20 * 0.9167:.4e}")
print(f"\nLook above at fTsumw or to_numpy().sum() for each file —")
print(f"  if those values match the documented POTs, bookkeeping is correct (deficit is real physics).")
print(f"  if NEW's value is closer to 7.14e20, the documented number is wrong.")
```
