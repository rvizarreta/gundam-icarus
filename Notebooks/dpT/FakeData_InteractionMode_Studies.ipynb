```python
import uproot
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

file_path = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/icarus_numi_numu_mc_onbeam_offbeam_syst_gundam_withPCA.root"

mode_labels = {0: "QE", 1: "RES", 2: "DIS", 3: "COH", 10: "MEC/2p2h"}
mode_colors = {0: "#4C72B0", 1: "#DD8452", 2: "#55A868", 3: "#C44E52", 10: "#9172B0"}

bins = np.linspace(0, 1.5, 20)  # GeV/c

with uproot.open(file_path) as f:
    for tree_name, label in [("events/full/selected", "Signal Selection"),
                              ("events/full/sideband",  "Sideband")]:

        tree   = f[tree_name]
        arrays = tree.arrays(["true_interaction_mode", "true_dpT_lp", "reco_dpT_lp"], library="np")

        modes    = arrays["true_interaction_mode"]
        true_dpt = arrays["true_dpT_lp"] / 1000.0   # MeV/c -> GeV/c
        reco_dpt = arrays["reco_dpT_lp"] / 1000.0   # MeV/c -> GeV/c

        unique_modes = np.unique(modes)
        plot_modes   = [int(m) for m in [0, 10, 1, 2, 3] if m in unique_modes]

        # ── 1. Interaction mode composition pie ──────────────────────────────
        fig, ax = plt.subplots(figsize=(6, 6))
        counts     = [int(np.sum(modes == m)) for m in plot_modes]
        pie_labels = [mode_labels[m] for m in plot_modes]
        pie_colors = [mode_colors[m] for m in plot_modes]
        ax.pie(counts, labels=pie_labels, colors=pie_colors, autopct="%1.1f%%", startangle=90)
        ax.set_title(f"Interaction Mode Composition — {label}", fontsize=13)
        plt.tight_layout()
        plt.show()

        # ── 2. δpT stacked histograms by interaction mode (true + reco) ──────
        for dpt, dpt_label in [(true_dpt, "True"), (reco_dpt, "Reco")]:
            fig, ax    = plt.subplots(figsize=(8, 5))
            # filter NaN/inf per mode
            data_list  = [dpt[(modes == m) & np.isfinite(dpt)] for m in plot_modes]
            stack_lbls = [mode_labels[m] for m in plot_modes]
            stack_clrs = [mode_colors[m] for m in plot_modes]
            ax.hist(data_list, bins=bins, stacked=True,
                    label=stack_lbls, color=stack_clrs, alpha=0.85,
                    edgecolor="black", linewidth=0.3)
            ax.set_xlabel(rf"{dpt_label} $\delta p_T$ [GeV/c]", fontsize=12)
            ax.set_ylabel("Events", fontsize=12)
            ax.set_title(rf"{dpt_label} $\delta p_T$ by Interaction Mode — {label}", fontsize=13)
            ax.legend(loc="upper right", fontsize=10)
            ax.xaxis.set_minor_locator(ticker.AutoMinorLocator())
            ax.yaxis.set_minor_locator(ticker.AutoMinorLocator())
            plt.tight_layout()
            plt.show()

        # nominal reco δpT histogram (used in plots 3 and 4) ─────────────────
        finite_mask = np.isfinite(reco_dpt)
        h_nom, _    = np.histogram(reco_dpt[finite_mask], bins=bins)

        # ── 3. Effect of the 0.8 RES reweight on reco δpT ────────────────────
        res_weight = np.where(modes == 1, 0.8, 1.0)
        h_rew, _   = np.histogram(reco_dpt[finite_mask], bins=bins, weights=res_weight[finite_mask])

        fig, axes = plt.subplots(2, 1, figsize=(8, 7),
                                 gridspec_kw={"height_ratios": [3, 1]},
                                 sharex=True)
        axes[0].step(bins, np.append(h_nom, h_nom[-1]), where="post",
                     color="black", label="Nominal")
        axes[0].step(bins, np.append(h_rew, h_rew[-1]), where="post",
                     color="red", linestyle="--", label=r"$0.8 \times$ RES")
        axes[0].set_ylabel("Events", fontsize=12)
        axes[0].set_title(rf"Effect of $0.8 \times$ RES reweight — {label}", fontsize=13)
        axes[0].legend(fontsize=10)

        with np.errstate(invalid="ignore", divide="ignore"):
            ratio_res = np.where(h_nom > 0, h_rew / h_nom, 1.0)
        axes[1].step(bins, np.append(ratio_res, ratio_res[-1]), where="post", color="red")
        axes[1].axhline(1.0, color="black", linestyle=":")
        axes[1].set_xlabel(rf"Reco $\delta p_T$ [GeV/c]", fontsize=12)
        axes[1].set_ylabel("Ratio", fontsize=12)
        axes[1].set_ylim(0.5, 1.1)
        for ax in axes:
            ax.xaxis.set_minor_locator(ticker.AutoMinorLocator())
            ax.yaxis.set_minor_locator(ticker.AutoMinorLocator())
        plt.tight_layout()
        plt.show()

        # ── 4. Effect of -1σ NormCCMEC (remove all MEC) on reco δpT ──────────
        mec_weight = np.where(modes == 10, 0.0, 1.0)
        h_mec, _   = np.histogram(reco_dpt[finite_mask], bins=bins, weights=mec_weight[finite_mask])

        fig, axes = plt.subplots(2, 1, figsize=(8, 7),
                                 gridspec_kw={"height_ratios": [3, 1]},
                                 sharex=True)
        axes[0].step(bins, np.append(h_nom, h_nom[-1]), where="post",
                     color="black", label="Nominal")
        axes[0].step(bins, np.append(h_mec, h_mec[-1]), where="post",
                     color="purple", linestyle="--", label=r"$-1\sigma$ NormCCMEC")
        axes[0].set_ylabel("Events", fontsize=12)
        axes[0].set_title(rf"Effect of $-1\sigma$ NormCCMEC — {label}", fontsize=13)
        axes[0].legend(fontsize=10)

        with np.errstate(invalid="ignore", divide="ignore"):
            ratio_mec = np.where(h_nom > 0, h_mec / h_nom, 1.0)
        axes[1].step(bins, np.append(ratio_mec, ratio_mec[-1]), where="post", color="purple")
        axes[1].axhline(1.0, color="black", linestyle=":")
        axes[1].set_xlabel(rf"Reco $\delta p_T$ [GeV/c]", fontsize=12)
        axes[1].set_ylabel("Ratio", fontsize=12)
        axes[1].set_ylim(0.5, 1.1)
        for ax in axes:
            ax.xaxis.set_minor_locator(ticker.AutoMinorLocator())
            ax.yaxis.set_minor_locator(ticker.AutoMinorLocator())
        plt.tight_layout()
        plt.show()
```
