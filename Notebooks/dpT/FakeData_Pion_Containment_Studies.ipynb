# Charged-Pion-Suppression (20%) Fake Data Study -- Containment Split

The fake data scales events with at least one true charged pion above the 25 MeV KE threshold (`true_no_charged_pions == 0`) by 0.8. These events are by definition background to the 1muNp0pi signal, not signal itself -- so unlike the RES/QE-suppression studies, there is no "suppressed truth" curve to overlay: the GENIE comparison below is nominal-only. A successful closure means the extraction returns the NOMINAL cross section despite the altered background, i.e. the fit correctly absorbs the pion-background mismodeling without biasing the signal measurement.
