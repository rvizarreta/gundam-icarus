```python
import uproot
import numpy as np
import matplotlib.pyplot as plt

path = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/icarus_numi_numu_mc_onbeam_offbeam_syst_gundam_withPCA.root"

f = uproot.open(path)
tree_sel = f["events/full/selected"]
tree_sb  = f["events/full/sideband"]

cat_sel  = np.asarray(tree_sel["category"].array())
q2_sel   = np.asarray(tree_sel["true_Q2"].array())
w_sel    = np.asarray(tree_sel["true_low_q2_weight"].array())

cat_sb   = np.asarray(tree_sb["category"].array())
q2_sb    = np.asarray(tree_sb["true_Q2"].array())
w_sb     = np.asarray(tree_sb["true_low_q2_weight"].array())

mask_sel = ((cat_sel == 6) | (cat_sel == 7) | (cat_sel == 8)) & ~np.isnan(w_sel) & (q2_sel > 0)
mask_sb  = ((cat_sb  == 6) | (cat_sb  == 7) | (cat_sb  == 8)) & ~np.isnan(w_sb)  & (q2_sb  > 0)

def stowell_weight(q2):
    Q2_max = 0.70
    Q2_t1  = 0.00
    Q2_t2  = 0.35
    R1     = 0.30
    R2     = 0.60
    w = np.ones_like(q2)
    valid = (q2 >= 0) & (q2 <= Q2_max)
    q = q2[valid]
    RQ2 = (R2 * ((q - Q2_t1) * (q - Q2_max)) / ((Q2_t2 - Q2_t1) * (Q2_t2 - Q2_max))
              + ((q - Q2_t1) * (q - Q2_t2))  / ((Q2_max - Q2_t1) * (Q2_max - Q2_t2)))
    w[valid] = 1.0 - (1.0 - R1) * (1.0 - RQ2)**2
    return w

# ── Plot 1 ────────────────────────────────────────────────────────────────────
bins = np.linspace(0, 2.0, 30)

h_sel, edges = np.histogram(q2_sel[mask_sel], bins=bins)
h_sb,  _     = np.histogram(q2_sb[mask_sb],  bins=bins)

fig, ax = plt.subplots(figsize=(7, 4))
ax.step(edges[:-1], h_sel / h_sel.sum(), where="post", label="signal selection (Other CC)")
ax.step(edges[:-1], h_sb  / h_sb.sum(),  where="post", label="sideband (Other CC)")
ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max = 0.70")
ax.set_xlabel("true Q2 [GeV^2]")
ax.set_ylabel("normalized area")
ax.set_title("true Q2 distribution - Other CC")
ax.legend()
fig.tight_layout()
save_fig(fig, r'.', 'plot1_q2_distribution')
plt.show()

# ── Plot 2 ────────────────────────────────────────────────────────────────────
q2_curve = np.linspace(0, 0.9, 500)
w_curve  = stowell_weight(q2_curve)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(q2_curve, w_curve)
ax.axvline(0.35, color="gray", linestyle=":",  linewidth=1.0, label="Q2_t2 = 0.35")
ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max = 0.70")
ax.axhline(0.30, color="green", linestyle="-.", linewidth=1.0, label="R1 = 0.30")
ax.set_xlabel("true Q2 [GeV^2]")
ax.set_ylabel("w(Q2)")
ax.set_title("Stowell low-Q2 suppression weight")
ax.legend()
fig.tight_layout()
save_fig(fig, r'.', 'plot2_stowell_weight')
plt.show()

# ── Plot 3 ────────────────────────────────────────────────────────────────────
bins_zoom = np.linspace(0, 1.5, 40)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, q2, w, mask, title in [
    (axes[0], q2_sel, w_sel, mask_sel, "signal selection - Other CC"),
    (axes[1], q2_sb,  w_sb,  mask_sb,  "sideband - Other CC"),
]:
    h_nom, edges = np.histogram(q2[mask], bins=bins_zoom)
    h_rw,  _     = np.histogram(q2[mask], bins=bins_zoom, weights=w[mask])

    ax.step(edges[:-1], h_nom, where="post", label="nominal")
    ax.step(edges[:-1], h_rw,  where="post", linestyle="--", label="low-Q2 suppressed")
    ax.fill_between(edges[:-1], h_rw, h_nom, step="post", alpha=0.3, label="suppressed events")
    ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max")
    ax.set_xlabel("true Q2 [GeV^2]")
    ax.set_ylabel("events")
    ax.set_title(title)
    ax.legend(fontsize=9)

fig.tight_layout()
save_fig(fig, r'.', 'plot3_nominal_vs_reweighted')
plt.show()
```

```python
"""
Checks whether the tree has BOTH 'true_Q2' and 'true_q_squared' branches
(Kiyoung's config uses the latter name), compares their distributions, and
tests which one gives physically sensible NuMI-scale Q^2 values.

Usage (terminal):
    python check_q2_branch_names.py /path/to/file.root

Usage (Jupyter):
    Edit FILE_PATH below and run the cell.
"""

import sys
import os
import numpy as np

try:
    import uproot
except ImportError:
    sys.exit("uproot is not installed. Run: pip install uproot numpy")

FILE_PATH = (
    "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/"
    "My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/"
    "ICARUS_CC0pi_Selection/data/icarus_numi_numu_mc_onbeam_offbeam_syst_ppfx.root"
)


def resolve_file_path():
    candidate = None
    if len(sys.argv) >= 2 and not sys.argv[1].startswith("-"):
        candidate = sys.argv[1]
    if candidate is None or not os.path.isfile(candidate):
        candidate = FILE_PATH
    if not os.path.isfile(candidate):
        sys.exit(f"No valid file found. Tried: {candidate!r}")
    return candidate


def summarize(name, arr):
    finite = arr[np.isfinite(arr)]
    n_nan = len(arr) - len(finite)
    print(f"    {name}: N={len(arr)}, NaN={n_nan}")
    if len(finite) == 0:
        return
    print(f"      min={finite.min():.6f}  max={finite.max():.6f}  "
          f"mean={finite.mean():.6f}  median={np.median(finite):.6f}")


def inspect_tree(file_path, tree_name):
    print(f"\n{'=' * 70}")
    print(f"Tree: {tree_name}")
    print(f"{'=' * 70}")

    f = uproot.open(file_path)
    if tree_name not in f:
        print(f"  Tree '{tree_name}' not found.")
        return
    tree = f[tree_name]
    branches = tree.keys()

    # Print EVERY branch whose name contains 'q' and a digit, or 'squared',
    # or 'q2' -- cast a much wider net than the earlier script did.
    candidates = sorted(set(
        b for b in branches
        if ("q2" in b.lower())
        or ("squared" in b.lower() and "q" in b.lower())
        or (b.lower().startswith("q_"))
        or (b.lower().endswith("_q"))
    ))
    print(f"  All Q2-like branch names found: {candidates}")

    has_Q2 = "true_Q2" in branches
    has_q_squared = "true_q_squared" in branches

    print(f"\n  'true_Q2' present:         {has_Q2}")
    print(f"  'true_q_squared' present:  {has_q_squared}")

    if not (has_Q2 or has_q_squared):
        print("  Neither branch found -- nothing further to check.")
        return

    load_list = []
    if has_Q2:
        load_list.append("true_Q2")
    if has_q_squared:
        load_list.append("true_q_squared")
    if "true_low_q2_weight" in branches:
        load_list.append("true_low_q2_weight")

    category_branch = "category" if "category" in branches else (
        "true_category" if "true_category" in branches else None
    )
    if category_branch:
        load_list.append(category_branch)

    arrs = tree.arrays(load_list, library="np")

    if has_Q2:
        print("\n  Distribution of true_Q2 (all events):")
        summarize("true_Q2", arrs["true_Q2"])

    if has_q_squared:
        print("\n  Distribution of true_q_squared (all events):")
        summarize("true_q_squared", arrs["true_q_squared"])

    if has_Q2 and has_q_squared:
        q2a = arrs["true_Q2"]
        q2b = arrs["true_q_squared"]
        both_finite = np.isfinite(q2a) & np.isfinite(q2b)
        n_both = both_finite.sum()
        if n_both > 0:
            identical = np.allclose(
                q2a[both_finite], q2b[both_finite], atol=1e-9, rtol=1e-6
            )
            print(f"\n  Are true_Q2 and true_q_squared identical where both "
                  f"finite (N={n_both})? {identical}")
            if not identical:
                ratio = q2a[both_finite] / np.where(
                    q2b[both_finite] == 0, np.nan, q2b[both_finite]
                )
                ratio = ratio[np.isfinite(ratio)]
                if len(ratio) > 0:
                    print(f"    true_Q2 / true_q_squared -- "
                          f"median ratio: {np.median(ratio):.4f}, "
                          f"mean ratio: {np.mean(ratio):.4f}")

    # If true_low_q2_weight and true_q_squared both exist, re-test the match
    # using Kiyoung's inline formula on true_q_squared instead of true_Q2.
    if has_q_squared and "true_low_q2_weight" in branches and category_branch:
        cat = arrs[category_branch]
        q2b = arrs["true_q_squared"]
        stored_w = arrs["true_low_q2_weight"]

        mask = np.isin(cat, [4, 5, 6, 7]) & np.isfinite(q2b) & np.isfinite(stored_w)
        n = mask.sum()
        if n > 0:
            q2_vals = q2b[mask]
            # Same Stowell formula, using true_q_squared as input this time.
            Q2_MAX, Q2_T2, R1, R2 = 0.70, 0.35, 0.30, 0.60
            w = np.ones_like(q2_vals)
            in_range = (q2_vals <= Q2_MAX) & (q2_vals >= 0.0)
            qv = q2_vals[in_range]
            term1 = -0.6 * qv * (qv - Q2_MAX) / 0.1225
            term2 = qv * (qv - Q2_T2) / 0.245
            RQ2 = term1 + term2
            w[in_range] = 1.0 - 0.7 * (1.0 - RQ2) ** 2

            diff = w - stored_w[mask]
            print(f"\n  Recomputing weight from true_q_squared instead of "
                  f"true_Q2 (categories 4-7, N={n}):")
            print(f"    max |recomputed - stored|: {np.max(np.abs(diff)):.6f}")
            n_below_max = np.sum((q2_vals >= 0.0) & (q2_vals <= Q2_MAX))
            print(f"    fraction of true_q_squared within [0, {Q2_MAX}]: "
                  f"{100.0 * n_below_max / n:.1f}%")


def main():
    file_path = resolve_file_path()
    for tree_name in ("events/full/selected", "events/full/sideband"):
        inspect_tree(file_path, tree_name)


if __name__ == "__main__":
    main()
```
