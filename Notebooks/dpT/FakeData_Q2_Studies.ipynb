```python
import uproot
import numpy as np
import matplotlib.pyplot as plt

path = "/Users/rvizarreta/Library/CloudStorage/GoogleDrive-rvizarreta14@gmail.com/My Drive/🏛 PhD Repository/🚀 Research/🤖 Experiments&Projects/ICARUS/ICARUS_CC0pi_Selection/data/icarus_numi_numu_mc_onbeam_offbeam_syst_gundam_withPCA.root"

f = uproot.open(path)
tree_sel = f["events/full/selected"]
tree_sb  = f["events/full/sideband"]

cat_sel  = np.asarray(tree_sel["category"].array())
q2_sel   = np.asarray(tree_sel["true_Q2"].array())
w_sel    = np.asarray(tree_sel["true_low_q2_weight"].array())

cat_sb   = np.asarray(tree_sb["category"].array())
q2_sb    = np.asarray(tree_sb["true_Q2"].array())
w_sb     = np.asarray(tree_sb["true_low_q2_weight"].array())

mask_sel = ((cat_sel == 6) | (cat_sel == 7) | (cat_sel == 8)) & ~np.isnan(w_sel) & (q2_sel > 0)
mask_sb  = ((cat_sb  == 6) | (cat_sb  == 7) | (cat_sb  == 8)) & ~np.isnan(w_sb)  & (q2_sb  > 0)

def stowell_weight(q2):
    Q2_max = 0.70
    Q2_t1  = 0.00
    Q2_t2  = 0.35
    R1     = 0.30
    R2     = 0.60
    w = np.ones_like(q2)
    valid = (q2 >= 0) & (q2 <= Q2_max)
    q = q2[valid]
    RQ2 = (R2 * ((q - Q2_t1) * (q - Q2_max)) / ((Q2_t2 - Q2_t1) * (Q2_t2 - Q2_max))
              + ((q - Q2_t1) * (q - Q2_t2))  / ((Q2_max - Q2_t1) * (Q2_max - Q2_t2)))
    w[valid] = 1.0 - (1.0 - R1) * (1.0 - RQ2)**2
    return w

# ── Plot 1 ────────────────────────────────────────────────────────────────────
bins = np.linspace(0, 2.0, 30)

h_sel, edges = np.histogram(q2_sel[mask_sel], bins=bins)
h_sb,  _     = np.histogram(q2_sb[mask_sb],  bins=bins)

fig, ax = plt.subplots(figsize=(7, 4))
ax.step(edges[:-1], h_sel / h_sel.sum(), where="post", label="signal selection (Other CC)")
ax.step(edges[:-1], h_sb  / h_sb.sum(),  where="post", label="sideband (Other CC)")
ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max = 0.70")
ax.set_xlabel("true Q2 [GeV^2]")
ax.set_ylabel("normalized area")
ax.set_title("true Q2 distribution - Other CC")
ax.legend()
fig.tight_layout()
save_fig(fig, r'.', 'plot1_q2_distribution')
plt.show()

# ── Plot 2 ────────────────────────────────────────────────────────────────────
q2_curve = np.linspace(0, 0.9, 500)
w_curve  = stowell_weight(q2_curve)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(q2_curve, w_curve)
ax.axvline(0.35, color="gray", linestyle=":",  linewidth=1.0, label="Q2_t2 = 0.35")
ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max = 0.70")
ax.axhline(0.30, color="green", linestyle="-.", linewidth=1.0, label="R1 = 0.30")
ax.set_xlabel("true Q2 [GeV^2]")
ax.set_ylabel("w(Q2)")
ax.set_title("Stowell low-Q2 suppression weight")
ax.legend()
fig.tight_layout()
save_fig(fig, r'.', 'plot2_stowell_weight')
plt.show()

# ── Plot 3 ────────────────────────────────────────────────────────────────────
bins_zoom = np.linspace(0, 1.5, 40)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, q2, w, mask, title in [
    (axes[0], q2_sel, w_sel, mask_sel, "signal selection - Other CC"),
    (axes[1], q2_sb,  w_sb,  mask_sb,  "sideband - Other CC"),
]:
    h_nom, edges = np.histogram(q2[mask], bins=bins_zoom)
    h_rw,  _     = np.histogram(q2[mask], bins=bins_zoom, weights=w[mask])

    ax.step(edges[:-1], h_nom, where="post", label="nominal")
    ax.step(edges[:-1], h_rw,  where="post", linestyle="--", label="low-Q2 suppressed")
    ax.fill_between(edges[:-1], h_rw, h_nom, step="post", alpha=0.3, label="suppressed events")
    ax.axvline(0.70, color="gray", linestyle="--", linewidth=1.0, label="Q2_max")
    ax.set_xlabel("true Q2 [GeV^2]")
    ax.set_ylabel("events")
    ax.set_title(title)
    ax.legend(fontsize=9)

fig.tight_layout()
save_fig(fig, r'.', 'plot3_nominal_vs_reweighted')
plt.show()
```
