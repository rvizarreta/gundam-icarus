# RES true-dpT SHAPE Fake Data Study -- Containment Split

Same RES true-delta_pT SHAPE closure check as `FakeData_RES_dpTShape_Studies.ipynb` (not yet built -- see the note below), but for the containment-split (contained/exiting) simultaneous fit. Reads from the `FakeData_RES_dpTShape_Containment` output directory.

**Fake data definition:** RES (`true_interaction_mode==1`) events are reweighted per true_dpT_lp_genie bin by the ratio of the GiBUU and GENIE true delta_pT shapes for RES-only events, instead of by a flat scale -- see `Configs_DataSetList/dataSetListConfig_FakeData_RES_dpTShape.yaml` for the derivation and the exact per-bin weights. This targets Smedley et al.'s CC>1p0pi multi-proton technote comment that GiBUU predicts a significantly more RES-dominated signal than the other generators, owing to its unique treatment of FSI -- checked (via the same GiBUU/GENIE delta_pT shape ratio used here) to be a genuine shape effect, not just a rate/composition effect.

**Baseline to compare against:** no combined-sample (non-containment) notebook exists for this study yet -- only the containment-split run below has been built, mirroring the precedent already set by `FakeData_RES_dpTShape_Studies.ipynb` not existing either (only its Containment sibling does).

**Binning:** truth binning is the medulla-optimized *combined* dpT binning (`binning_true_dpT.txt`, 6 finite + 1 overflow = 7 bins, shared across all 4 samples -- the same 7 bins the RES dpT-shape reweight itself is defined on). Reco binning is per-channel: contained samples use `binning_reco_dpT_contained.txt` (20 finite + 1 overflow = 21 rows), exiting samples use `binning_reco_dpT_exiting.txt` (10 finite + 1 overflow = 11 rows).

**`cov_bin_range` -- carried over unchanged from the RES_dpTShape_Containment closure (itself inherited from the original QE_Containment closure test), NOT yet re-verified for this run.** Block boundaries in the full 64x64 matrix are at 0, 21, 32, 53, 64 (with-overflow widths); each slice used for chi2 stops one row short of its block boundary to match the category histograms: selection_contained (0,20), selection_exiting (21,31), sideband_contained (32,56), sideband_exiting (57,81) -- the sideband blocks are wider than the selection blocks because the sideband uses its own dedicated 24-finite-bin binning, not the selection's. Since `enablePca: true` reduces parameters based on the actual response matrix, and this RES dpT-shape reweighting (RES is the dominant mode in the pi+/- sideband, unlike QE) could shift bin population very differently than the QE study, re-check these boundaries against this run's own `covarianceMatrix_TH2D` axis labels before trusting the per-sample chi2 breakdown.
