# Muon Angle Asimov Containment Closure Test

Closure check for the containment-split (contained/exiting) simultaneous fit in the muon angle with respect to the NuMI beam, $\cos\theta_\mu$. It reads the `Asimov_Containment` outputs of the fitter, cross-section and toy-generator runs listed in `RunConfigs/muon_angle/Asimov_Containment/Asimov_Containment_instructions.md`. The structure follows `opening_angle/Asimov_Containment_Studies.ipynb`.

**Binning:** truth binning is `binning_true_muon_angle.txt` (7 bins from -1 to 1, shared across all 4 samples). It uses every second edge of the 14-bin combined reco binning, taken from `leading_muon_NuMI_angle.toml` in the selection repository. Reco binning is per channel: contained selection uses `binning_reco_leading_muon_angle_contained.txt` (18 bins), exiting selection uses `binning_reco_leading_muon_angle_exiting.txt` (14 bins), and both sidebands use their own 24-bin files. $\cos\theta_\mu$ is bounded to [-1, 1], so none of these binnings has an overflow bin and no bin is dropped when plotting (`remove_last_bin=False` in the scatter plots).

**`cov_bin_range` -- NOT YET VERIFIED against a muon-angle output file.** With no overflow bin, each sample's block in the full covariance matrix is expected to have the width of its binning. The full matrix is then 18 + 14 + 24 + 24 = 80 rows and the slices are (0,18), (18,32), (32,56), (56,80). The check cell below compares these widths against the bin counts of the category histograms and the shape of `covarianceMatrix_TH2D` in the actual file. Run it first and edit `COV_RANGE` if it reports a mismatch.

## Binning Sanity Check

Compares the expected covariance block widths in `COV_RANGE` with the actual contents of the pre-fit toy-generator file. Every line should end in `OK`.

```python
import uproot

f_pre = uproot.open(PREFIT_FILE)
cov_full = f_pre['toyGen/matrices/covarianceMatrix_TH2D'].values()
expected_rows = max(hi for _, hi in COV_RANGE.values())
print(f'covarianceMatrix_TH2D shape: {cov_full.shape}  (expected {expected_rows} x {expected_rows})  ->',
      'OK' if cov_full.shape == (expected_rows, expected_rows) else 'MISMATCH')

for key, sample in SAMPLES.items():
    lo, hi = COV_RANGE[key]
    nbins = None
    for cat in range(11):
        path = f'toyGen/plots/histograms/{sample} (pre-fit)/reco_leading_muon_cos_theta_mu/category/{cat}/MC_TH1D'
        try:
            nbins = len(f_pre[path].values())
            break
        except KeyError:
            continue
    status = 'OK' if nbins == hi - lo else 'MISMATCH'
    print(f'{sample:<36} category-histogram bins: {nbins}  |  COV_RANGE width: {hi - lo}  ->  {status}')
```
